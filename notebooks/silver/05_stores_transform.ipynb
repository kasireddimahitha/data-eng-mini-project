```python
from pyspark.sql.functions import col, lower, trim

df = spark.read.table("automobile_catalog.001_bronze.store")

df = df.dropDuplicates()

df = df.withColumn("store_name", lower(trim(col("store_name")))) \
       .withColumn("city", lower(trim(col("city")))) \
       .withColumn("state", lower(trim(col("state"))))

df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("automobile_catalog.002_silver.store")
```
