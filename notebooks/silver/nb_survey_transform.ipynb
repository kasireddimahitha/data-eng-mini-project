```python
from pyspark.sql.functions import col, expr

df = spark.read.table("automobile_catalog.001_bronze.customer_survey")

df = df.dropDuplicates()

df = df.withColumn(
    "survey_sent_date",
    expr("to_date(try_to_timestamp(survey_sent_date, 'dd-MM-yyyy HH:mm'))")
).withColumn(
    "survey_response_date",
    expr("to_date(try_to_timestamp(survey_response_date, 'dd-MM-yyyy HH:mm'))")
)

df = df.fillna({"overall_satisfaction_rating": 0})

df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("automobile_catalog.002_silver.customer_survey")
```
