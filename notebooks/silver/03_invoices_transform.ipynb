```python
from pyspark.sql.functions import col, expr

df = spark.read.table("automobile_catalog.001_bronze.invoice")

df = df.dropDuplicates()

# FIX: handle dd-MM-yyyy HH:mm format
df = df.withColumn(
    "invoice_date",
    expr("to_date(try_to_timestamp(invoice_date, 'dd-MM-yyyy HH:mm'))")
)

df = df.withColumn("invoice_amount", col("invoice_amount").cast("double"))

df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("automobile_catalog.002_silver.invoice")
```
