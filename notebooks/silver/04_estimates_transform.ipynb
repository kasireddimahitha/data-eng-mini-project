```python
from pyspark.sql.functions import col, expr

df = spark.read.table("automobile_catalog.001_bronze.estimate")

df = df.dropDuplicates()

df = df.withColumn(
    "created_at",
    expr("try_to_timestamp(created_at, 'dd-MM-yyyy HH:mm')")
)

df = df.withColumn("estimate_amount", col("estimate_amount").cast("double"))

df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("automobile_catalog.002_silver.estimate")
```
