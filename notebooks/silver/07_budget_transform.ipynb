```python
from pyspark.sql.functions import col, expr

df = spark.read.table("automobile_catalog.001_bronze.ns_budget")

df = df.dropDuplicates()

df = df.withColumn(
    "month",
    expr("to_date(try_to_timestamp(month, 'dd-MM-yyyy HH:mm'))")
)

df = df.withColumn("budget_amount", col("budget_amount").cast("double"))

df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("automobile_catalog.002_silver.ns_budget")
```
