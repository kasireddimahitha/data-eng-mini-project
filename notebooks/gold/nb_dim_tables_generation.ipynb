```sql
-- Create the dim_store table in the gold layer
CREATE OR REPLACE TABLE automobile_catalog.003_gold.dim_store AS
SELECT DISTINCT
    store_id,
    store_name,
    city,
    state,
    manager_id,
    manager_name,
    opened_year,
    store_type
FROM automobile_catalog.002_silver.store;
select *from automobile_catalog.003_gold.dim_store;
```

```sql
-- Create or replace the dim_technician table in the gold layer
CREATE OR REPLACE TABLE automobile_catalog.003_gold.dim_technician AS
SELECT DISTINCT
    o.technician_id,
    o.technician_name,
    s.store_id,
    s.store_name
FROM automobile_catalog.002_silver.order AS o
INNER JOIN automobile_catalog.002_silver.store AS s
    ON o.store_id = s.store_id;
select *from automobile_catalog.003_gold.dim_technician;
```

```sql
-- Create or replace dim_estimator table in the gold layer
CREATE OR REPLACE TABLE automobile_catalog.003_gold.dim_estimator AS
SELECT DISTINCT
    estimator_id,
    estimator_name
FROM automobile_catalog.002_silver.estimate;
select *from automobile_catalog.003_gold.dim_estimator;
```
