```python
 
files = [
    'customer_survey.csv',
    'estimate.csv',
    'invoice.csv',
    'ns_budget.csv',
    'order.csv',
    'store.csv'
]
 
 
s3_base_path = 's3://my-company-dms-postgres/'
 
for file in files:
    table_name = file.replace('.csv', '')
   
    df = spark.read.csv(s3_base_path + file, header=True, inferSchema=True)
   
    target_table = f"automobile_catalog.001_bronze.{table_name}"
    df.write.mode("overwrite").saveAsTable(target_table)
   
 
print("\n All files loaded to bronze schema successfully!")
 
 
```
